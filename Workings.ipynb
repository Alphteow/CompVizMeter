```python
import cv2
import pytesseract
from pytesseract import Output
import os
import re
import numpy as np
import tempfile
from PIL import Image
```

```python
# Define input and output directories
input_folder = 'Data'
output_folder = 'Grayscale'

# Create the output folder if it doesn't exist
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Loop through all files in the input folder
for filename in os.listdir(input_folder):
    if filename.endswith(('.png', '.jpg', '.jpeg')):  # Check for image files
        # Read the image
        img_path = os.path.join(input_folder, filename)
        image = cv2.imread(img_path)

        # Convert the image to grayscale
        gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        # Get the base filename without extension
        base_filename = os.path.splitext(filename)[0]

        # Save the grayscale image as PNG to the output folder
        output_filename = base_filename + '.png'
        output_path = os.path.join(output_folder, output_filename)
        cv2.imwrite(output_path, gray_image)

print("Conversion to grayscale and saving as PNG completed!")
```

```python
def analyze_image(image_path):
    # Read the grayscale image
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    # Calculate the mean pixel intensity
    mean_intensity = np.mean(image)

    # Calculate the standard deviation of pixel intensities
    std_intensity = np.std(image)

    # Define thresholds for over/underexposure
    overexposed_threshold = 200
    underexposed_threshold = 50

    # Check for overexposure and underexposure
    overexposed = np.sum(image > overexposed_threshold) / image.size
    underexposed = np.sum(image < underexposed_threshold) / image.size

    return mean_intensity, std_intensity, overexposed, underexposed

# Directory containing the grayscale images
grayscale_dir = 'Grayscale'

# Iterate over the grayscale images
for filename in os.listdir(grayscale_dir):
    if filename.endswith('.png'):
        image_path = os.path.join(grayscale_dir, filename)
        mean, std, over, under = analyze_image(image_path)
        print(f"Image: {filename}")
        print(f"Mean Intensity: {mean:.2f}")
        print(f"Standard Deviation: {std:.2f}")
        print(f"Overexposed Percentage: {over:.2%}")
        print(f"Underexposed Percentage: {under:.2%}")
        print("---")
```
